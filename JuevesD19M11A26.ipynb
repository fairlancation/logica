Bono: zip()

```python
lista1 = [1,2,3,4]
lista2 = ["a","b","c","d"]
for e1, e2 in zip(lista1, lista2):
  print(e1*e2)
```

Libreria iber tools Count() Genera start, start + setp, ... para siempre

```python
from itertools import count
for i in count(10,3):
  if i > 20: break
  print(i)
```

Cycle(iterable) reoute el iterable indefinidamente

```python
from itertools import cycle
colores = cycle(['rojo', 'verde', 'azul'])
#[next(colores) for _ in range(7)]
while True:
  print(next(colores))
  aburrido = input()
  if aburrido == 'si': break
```

repeat(obj, times = None)
Repite un objeto. Sin times es infinito

```python
from itertools import repeat
alo = repeat('x', 4)
next(alo)
```

product(*iterables, repeat = 1)
producto cartesiano. Rela del producto: len(a)*len(b)

```python
from itertools import product
list(product('AB', '12'))
list(product('AB', repeat=2))
list(product([1,2,3],[4,5,6],[7,8,9]))
```

permutaciones(iterable, r= None)
orden importa, sin repeticion, cantidad: P(n,r) = n!/(n-r)
combinaciones

```python
from itertools import permutations as permu
list(permu([1,2,3,4,5,6,7], 3))
from itertools import combinations
list(combinations([1,2,3,4,5,6,7], 3))
```

Ejercicio 2: atacar una contraseña

Se sabe que tiene las letras a,b,z,n,y; ademas de los numeros 1,5,6

Se sabe tambien que la contraseña tiene 5 caracteres de longitud

```python
from itertools import permutations as permu
caracteres = ['a','b','z','n','y', 1,5,6]
list(permu([caracteres], 5))
```
